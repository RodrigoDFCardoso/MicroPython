```python
import matplotlib.pyplot as plt

# Dados da bateria
tensao_bateria = 3.7  # Tensão nominal da bateria em volts
capacidade_bateria = 2600  # Capacidade da bateria em mAh
resistores = [10.0, 4.7, 3.3]  # Resistências em ohms

# Cálculo da capacidade em Ah
capacidade_bateria_Ah = capacidade_bateria / 1000

# Cria listas para armazenar os dados da descarga
tempo = []
corrente = []

# Simulação de descarga para cada resistor
for resistor in resistores:
    corrente_atual = tensao_bateria / resistor  # Corrente em A
    capacidade_restante = capacidade_bateria_Ah

    # Simule a descarga até que a capacidade da bateria seja considerada totalmente descarregada
    while capacidade_restante > 0:
        tempo.append(len(tempo))  # Tempo em segundos
        corrente.append(corrente_atual)  # Corrente em A
        capacidade_restante -= (corrente_atual * 1)  # Atualize a capacidade com base no tempo

# Criação do gráfico
plt.figure(figsize=(10, 6))
for i, resistor in enumerate(resistores):
    plt.plot(tempo[i::len(resistores)], corrente[i::len(resistores)], label=f"Resistor {resistor} ohms")

plt.title("Descarga da Bateria com Diferentes Resistores")
plt.xlabel("Tempo (s)")
plt.ylabel("Corrente (A)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dados da bateria
tensao_nominal_bateria = 3.7  # Tensão nominal da bateria em volts
tensao_corte_bateria = 3.0  # Tensão de corte da bateria em volts
capacidade_bateria = 2600  # Capacidade da bateria em mAh
resistores = [10.0, 4.7, 3.3]  # Resistências em ohms

# Cálculo da capacidade em Ah
capacidade_bateria_Ah = capacidade_bateria / 1000

# Criação de listas para armazenar os dados de cada cenário
tempo_resistores = []
tensao_resistores = []

# Simulação de descarga para cada cenário de resistência
for resistor in resistores:
    tensao_nominal_bateria_atual = tensao_nominal_bateria
    tempo_atual = 0
    tensao_atual = []

    # Simule a descarga até a tensão de corte da bateria ser atingida
    while tensao_nominal_bateria_atual > tensao_corte_bateria:
        # Adicione os dados atuais às listas
        tempo_atual += 1  # Incrementa o tempo em 1 segundo
        tensao_atual.append(tensao_nominal_bateria_atual)  # Registra a tensão atual

        # Calcule a corrente com base na Lei de Ohm
        corrente_atual = tensao_nominal_bateria_atual / resistor

        # Atualize a tensão com base na corrente
        tensao_nominal_bateria_atual -= (corrente_atual * 1)

    tempo_resistores.append(list(range(tempo_atual)))
    tensao_resistores.append(tensao_atual)

# Criação do gráfico
plt.figure(figsize=(10, 6))

for i, resistor in enumerate(resistores):
    plt.plot(tempo_resistores[i], tensao_resistores[i], label=f"Resistor {resistor} ohms")

plt.title("Descarga da Bateria até Tensão de Corte")
plt.xlabel("Tempo (s)")
plt.ylabel("Tensão da Bateria (V)")
plt.grid(True)
plt.legend()
plt.show()
```
